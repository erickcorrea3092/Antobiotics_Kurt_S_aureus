## Clases

Python es un [lenguaje orientado a objetos](https://en.wikipedia.org/wiki/Object-oriented_programming). Un objeto es un espacio en la memoria que cuenta con **comportamiento**, **estado** e **identidad**.

* Comportamiento: Representado por métodos o funciones tales como `dormir()`, `correr()`, `saludar()` etc.
* Estado: Representado por atributos tales como `edad`, `nombre`, etc.
* Identidad: Al crear los objetos obtenemos una referencia del objeto para poder acceder a su estado y comportamiento.

```
class Persona(object):
  # constructor. cuando se crea un nuevo objeto se llamara a esta funcion
  def __init__(self, nombre, edad):
    # self.nombre, self.edad y self.energia son atributos del objeto persona
    self.nombre = nombre
    self.edad = edad
    self.energia = 10

  # dormir es un método de los objetos persona. self es el objeto persona
  def dormir(self):
    self.energia = 10
    print('zzz')

  def correr(self):
    if self.energia >= 5:
      print('Corriendo...')
      self.energia -= 2
    else:
      print('Estoy muy cansado para correr :(')

  def saludar(self):
    print('Hola, me llamo {} y tengo {} años'.format(self.nombre, self.edad))
```

Para instanciar un objeto necesitaremos acceder a la clase `Persona` y la llamamos con los parámetros que querramos.

```
pepe = Persona('pepe', 22)
```

```

pepe.saludar()
pepe.correr() #10 corriendo
```

```
pepe.correr()  #8 corriendo
pepe.correr()#6 corriendo
pepe.correr()#4 Ya no corre
```

```



pepe.dormir()#4 Se recupera Energía = 10
pepe.correr() # 10 corriendo
```

Además en los lenguajes orientados a objetos existe un término conocido como herencia [herencia](https://en.wikipedia.org/wiki/Inheritance_(object-oriented_programming)). La herencia sirve para reutilizar comportamiento de una clase existente (como `Persona`) y expandirlo para que pueda hacer cosas más específicas, dependiendo de lo que se quiera resolver.

Por ejemplo, podemos definir una clase `Trabajador` que haga todo lo que pueda hacer una persona pero aparte pueda trabajar.

```
# clase trabajador. De esta manera sabemos que hereda de la clase Persona
class Trabajador(Persona):
  # sobreescribimos el constructor para agregar la propiedad lugar_de_trabajo
  def __init__(self, nombre, edad, lugar_de_trabajo):
    # llamamos al constructor de Persona, este inicializa nombre y edad
    super().__init__(nombre, edad)
    self.lugar_de_trabajo = lugar_de_trabajo

  def trabajar(self):
    # dado que Trabajador hereda de Persona, puede acceder al atributo energia
    if self.energia >= 5:
      print('Estoy trabajando en {}'.format(self.lugar_de_trabajo))
    else:
      print('Estoy muy cansado para trabajar :(')
```

Ya existía un método constructor en la clase padre, sin embargo en `Trabajador` estamos definiendo un nuevo constructor. Esto es conocido como [sobreescritura de métodos](https://en.wikipedia.org/wiki/Method_overriding). Esta sobreescritura nos permite definir los métodos nuevamente para cambiar el comportamiento de funciones existentes, inicializando el atributo `lugar_de_trabajo` en este caso. La llamada a la función [`super`](https://docs.python.org/3/library/functions.html#super) nos permite correr métodos de instancia en el contexto de la clase padre (`Persona`); entonces, al llamar al constructor desde `super` ejecutará el constructor definido en `Persona`. Es por esto que se inicializan los atributos `nombre`, `edad` y `energia`. De este modo, `Trabajador` puede hacer todo lo que puede hacer `Persona`  (`saludar`, `dormir` y `correr`) pero también puede trabajar, cosa única para él.

```
pedro = Trabajador('pedro', 31, 'Walmart')
```

```
pedro.correr()#10-2
```

```

pedro.trabajar() #8
pedro.dormir() #10
pedro.correr() #10-2
pedro.correr() #8-2
pedro.correr()#6-2 = 4
pedro.trabajar()
pedro.dormir() #10
pedro.trabajar()
```

## Condicionales

Hay veces que el comportamiento de nuestro programa cambiará dependiendo de los valores de algunas variables. Para determinar este flujo se utilizarán los condicionales `if`, `elif` y `else`.

```
0.1*0.1*3.1415
```

```
def determinar_area_apropiada(radio):

  area = calcular_area_circulo(radio)
  if area < 1:
    return 'Círculo muy pequeno'
  elif area < 10:
    return 'Círculo mediano'
  else:
    return 'Circulo grande'

# 3.1415*radio*radio
print(determinar_area_apropiada(0.1)) #círculo pequeño
print(determinar_area_apropiada(1)) #círculo mediano
print(determinar_area_apropiada(10)) #círculo grande
```

## Ciclos

```
radios = [ 1, 2, 3, 4, 5, 6 ]
radios
```

Ahora, vamos a calcular los volúmenes de las esferas para cada radio en la lista `radios`.

```
def calcular_volumen_esfera(radio):
  # Volumen de una esfera = (4/3) * pi * r^3
  # Que también se puede expresar como (4/3) * (pi * r^2) * r
  # donde pi * r^2 es el área del círculo.
  area = calcular_area_circulo(radio)
  return (4/3) * area * radio

print('Volúmenes de esferas para los radios en la lista:')
for r in radios:
  volumen = calcular_volumen_esfera(r)
  print(f'Radio: {r}, Volumen: {volumen}')
```

## Listas por comprensión

Es una forma de crear listas mas facilmente en una sola linea

```
# original
l1 = list()
for i in range(20):
  if i % 2 == 0:  # si i es par
    l1.append(i)

# lista por comprension
l2 = [ i for i in range(20) if i % 2 == 0 ]

print(l1)
print(l2)
```

## Tuplas

Las tuplas nos permiten almacenar varios valores en una sola variable. Muy util para funciones que necesiten regresar mas de un valor.

```
tupla = (10, 'a')
print(tupla, type(tupla))

# regresa el maximo y su indice de un iterable
def max_y_argmax(iterable):
  max_val = -float('inf')
  max_arg = 0
  for i in range(len(iterable)):
    if iterable[i] > max_val:
      max_val = iterable[i]
      max_arg = i
  return max_val, max_arg

l = [ 1, 2, 3, 4, 5, 4, 3, 2, 1 ]
val, arg = max_y_argmax(l) # se pueden agarrar los valores por separado
print(val, arg)
```

## Zip

Permite iterar dos o mas iterables al mismo tiempo

```
nums = [ 1, 2, 3, 4, 5, 6 ]
cadenas = [ 'a', 'b', 'c', 'd', 'e', 'f' ]

for num, cadena in zip(nums, cadenas):
  print(num, cadena)
```

## Muchas más

Python ofrece muchos otros atajos para poder escribir codigos muy rapidamente. Estos atajos suelen hacer mas facil la tarea de programar, pero también suele hacer los códigos más difíciles de entender!

Por ejemplo, podemos transformar todo este bloque:

```
th = 25
x1 = np.random.rand(100, 1000)*2 - 1
x2 = np.random.rand(100, 1000)*2 - 1

def d(x, y):
  return np.linalg.norm(x-y)

arr = list()
for i in range(len(x1)):
  dist = d(x1[i], x2[i])
  if abs(dist) > th:
    arr.append((i, dist))
  else:
    arr.append((i, float('inf')))

def comparador(tupla):
  return tupla[1]

arr = sorted(arr, key=comparador)
print(arr[:10])
```

En esto:

```
sorted([ (i, x) if abs(x) > th else (i, float('inf')) for i, x in
    enumerate(map(lambda t: np.linalg.norm(t[0]-t[1]), zip(x1, x2))) ],
    key=lambda t: t[1])[:10]
```
